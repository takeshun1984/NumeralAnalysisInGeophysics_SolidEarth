関数$f(x)$の2次精度中央差分の式は、
$$
\frac{df(x)}{dx} = \frac{f(x+\Delta x)-f(x-\Delta x)}{2\Delta x}
$$
となる。以下、解析的に微分が可能な$f(x)=e^x\sin{(x)}$を用いて、$f'(x)=e^x(\sin{(x)}+\cos{(x)})$と2次精度中央微分による導関数$f'_{FD}(x)$を比較する。

````python
import numpy as np

# x範囲を設定
x = np.linspace(-2, 2, 401)
dx = x[1] - x[0]
print(f"dx : {dx:.4f}")

# 関数
y = np.exp(x) * np.sin(x)

# 理論
dxy = np.exp(x)*(np.sin(x)+np.cos(x)) 

# 中央差分（2次精度）
dxy_fd = np.zeros_like(y)
dxy_fd[1:-1] = (y[2:] - y[:-2]) / (2 * dx)

# for文版
# for i in range(1, n-1):
#    dxy_fd[i] = (y[i+1] - y[i-1]) / (2 * dx)

# 境界は片側差分で代替
# 左端 (index 0)
dxy_fd[0] = (y[1] - y[0]) / dx
# 右端 (index -1)
dxy_fd[-1] = (y[-1] - y[-2]) / dx
````

解析的に微分した結果dxyと、差分を用いた結果dxy_fdを図示して比較してみる。

````python
import matplotlib.pyplot as plt

# === プロット ===
plt.rcParams.update({'font.size': 10}) #全体のフォントサイズ
plt.rcParams['xtick.direction'] = 'in' #x軸の目盛りの向き
plt.rcParams['ytick.direction'] = 'in' #y軸の目盛りの向き
plt.rcParams["xtick.minor.visible"] = True  #x軸補助目盛りの追加
plt.rcParams["ytick.minor.visible"] = True  #y軸補助目盛りの追加


plt.figure(figsize=(4,4))
plt.plot(x, y,color='black',linewidth=1,label=r'$y = e^x \sin(x)$')
plt.plot(x, dxy,color='lightblue',linewidth=4,label='Analytical')
plt.plot(x, dxy_fd,color='blue',linewidth=1,linestyle='--',label='FD')
plt.xlabel("x")
plt.grid(True)
plt.legend(loc='upper left')
plt.xlim(-2,2)
plt.show()
````

````python
# 誤差の計算
# 両端は1次精度の片側差分なので、中央差分の誤差を見るため内部点のみ描画する。
diff_abs = np.abs(dxy_fd - dxy)              # 絶対誤差
diff_rel = np.abs((dxy_fd - dxy) / dxy) * 100 # 相対誤差 [%]
# 注意：相対誤差は f'(x)=0 となる x=-π/4 付近で分母が0に近づき大きくなるため、対数軸で描画する。

# === プロット ===
plt.rcParams.update({'font.size': 10}) #全体のフォントサイズ
plt.rcParams['xtick.direction'] = 'in' #x軸の目盛りの向き
plt.rcParams['ytick.direction'] = 'in' #y軸の目盛りの向き
plt.rcParams["xtick.minor.visible"] = True  #x軸補助目盛りの追加
plt.rcParams["ytick.minor.visible"] = True  #y軸補助目盛りの追加


fig, axs = plt.subplots(1, 2, figsize=(8,4))

ax = axs[0]
ax.plot(x[1:-1], diff_abs[1:-1],color='blue',linewidth=1,linestyle='-')
ax.set_ylabel("Absolute error")

ax = axs[1]
ax.semilogy(x[1:-1], diff_rel[1:-1],color='red',linewidth=1,linestyle='-')
ax.axvline(-np.pi/4, color='gray', linestyle='--', linewidth=1, label=r"$f'(x)=0$")
ax.set_ylabel("Relative error [%]")
ax.legend(loc='upper right')

for ax in axs:
    ax.set_xlabel("x")
    ax.grid(True)
    ax.set_xlim(-2,2)

plt.tight_layout()
plt.show()
````

左図は絶対誤差 $|f'_{FD}(x)-f'(x)|$、右図は相対誤差 $|f'_{FD}(x)-f'(x)|/|f'(x)|$ を示している。

数値計算では誤差の大きさを真値との比で評価する相対誤差が重要である。ただし、相対誤差には**発散箇所**があることに注意が必要である。$f'(x)=e^x(\sin{(x)}+\cos{(x)})$ は $x=-\pi/4$ でゼロとなるため、相対誤差の分母が0に近づき、この付近で相対誤差が急激に大きくなる（右図の破線）。これは差分法の精度が悪化したわけではなく、左図の絶対誤差はこの付近でも小さいままである。真値がゼロに近い点では、相対誤差ではなく絶対誤差で評価する必要がある。

また、$x=\pi/4$ 付近で誤差が小さくなるのは、2次精度中央差分の誤差の主要項 $\frac{\Delta x^2}{6}f'''(x)$ において $f'''(x)=2e^x(\cos{(x)}-\sin{(x)})$ がゼロとなるためである。
